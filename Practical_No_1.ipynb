```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load dataset
df = pd.read_csv('retail_sales_dataset.csv')

# Histogram Plot - Age Group Distribution
plt.figure(figsize=(20, 5))
sns.histplot(data=df, x='Age', kde=True, color='c')
plt.title('Age Group Distribution')
plt.xticks(rotation=5)
plt.show()
```

```python
# Hist plot - Gender Group Distribution
plt.figure(figsize=(20, 5))
sns.histplot(data=df, x=df['Gender'].sort_values(), color='g')
```

```python
import plotly.express as px

# Violin plot - Gender & Age Group Distribution
plot = px.violin(df, x='Gender', y='Age', box=True, points='all')
plot.update_traces(box_fillcolor='green', opacity=0.5)
plot.update_layout(
    title='Gender & Age Group Distribution',
    xaxis_title='Gender',
    yaxis_title='Age Group'
)
plot.show()
```

```python
# Bar Plot
Gold_item_net_wt = df.groupby('Product Category')['Total Amount'].sum().nlargest(40)

plt.figure(figsize=(12, 6))
sns.barplot(x=Gold_item_net_wt.index, y=Gold_item_net_wt.values)
plt.xticks(rotation=90)
plt.title("Product Category -- Total Amount")
plt.xlabel("Product Category")
plt.ylabel("Total Amount")
plt.show()
```

```python
# Bar Plot
Gold_item_net_wt = df.groupby('Product Category')['Total Amount'].sum().nlargest(40)

plt.figure(figsize=(12, 6))
sns.barplot(x=Gold_item_net_wt.index, y=Gold_item_net_wt.values)
plt.xticks(rotation=90)
plt.title("Product Category -- Total Amount")
plt.xlabel("Product Category")
plt.ylabel("Total Amount")
plt.show()
```

```python
# Bar Plot - Quantity sold per Product Category
Gold_item_net_wt = df.groupby('Product Category')['Quantity'].sum().nlargest(40)

plt.figure(figsize=(12, 6))
sns.barplot(x=Gold_item_net_wt.index, y=Gold_item_net_wt.values)
plt.xticks(rotation=90)
plt.title("Product Category -- Total Quantity")
plt.xlabel("Product Category")
plt.ylabel("Quantity")
plt.show()
```
